# Автоматическая классификация тональности RuReviews

Ноутбук представляет сохранённые результаты воспроизводимого эксперимента `src/run_experiment.py`. Ячейки читают файлы из `results/` и `figures/`.

```
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

ROOT = Path('.').resolve()
RESULTS = ROOT / 'results'
FIGURES = ROOT / 'figures'
print('ROOT:', ROOT)
print('results exists:', RESULTS.exists())
```

## 1. Данные

```
summary = json.loads((RESULTS / 'data_summary.json').read_text(encoding='utf-8'))
summary
```

```
display(Image(filename=str(FIGURES / 'class_distribution.png')))
display(Image(filename=str(FIGURES / 'text_length_distribution.png')))
```

## 2. Сравнение моделей

```
metrics = pd.read_csv(RESULTS / 'model_metrics.csv')
metrics
```

```
display(Image(filename=str(FIGURES / 'model_comparison.png')))
```

## 3. Матрица ошибок и признаки

```
display(Image(filename=str(FIGURES / 'confusion_matrix.png')))
display(Image(filename=str(FIGURES / 'top_features.png')))
pd.read_csv(RESULTS / 'classification_report.csv', index_col=0)
```

## 4. Калибровка

```
calibration = json.loads((RESULTS / 'calibration_metrics.json').read_text(encoding='utf-8'))
display(Image(filename=str(FIGURES / 'calibration_curve.png')))
calibration
```

## Вывод

Лучшая модель: **LogisticRegression**, macro-F1 = 0,750, ROC-AUC = 0,898.
